## Question 6:  Median value of horsepower 
Answer: Yes, it decreased

```python
median_before = car_df['horsepower'].median()

most_frequent = car_df['horsepower'].mode()[0]
car_df['horsepower'] = car_df['horsepower'].fillna(most_frequent)

median_after = car_df['horsepower'].median()

print(median_before, most_frequent, median_after)
```

## Question 7:  Sum of weights 
Answer: 0.369

```python
X = car_df[car_df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7).values

XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

w = XTX_inv @ X.T @ y
w.sum()
```
