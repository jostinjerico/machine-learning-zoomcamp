```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

Data Preparation

```python
car_df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
```

```python
car_df.head()
```

Renaming the columns

```python
car_df.columns = car_df.columns.str.lower().str.replace(' ', '_')
```

```python
car_df.head()
```

```python
car_df.dtypes
```

```python
strings = list (car_df.dtypes[car_df.dtypes == 'object'].index)
```

```python
for col in strings:
    car_df[col] = car_df[col].str.lower().str.replace(' ', '_')
```
